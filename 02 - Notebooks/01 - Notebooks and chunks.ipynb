
# Notebooks

One of the key benefits of using notebooks is that they have both *code chunks* and *markdown chunks* which can be used to document your process as you go. This text is stored in a _markdown chunk_. If you double click this text you'll find yourself able to edit it.

There are **several** advantages to this over using code comments:
- Formatting of text makes it far easier for a person to read
- You can include images and diagrams in markdown chunks to convey complexity far more easily than trying to describe it in text
- Markdown headers can be used to clearly separate out different sections of the notebook
- Markdown headers also allow you to navigate the document more easily

Notebooks are compatible with SQL warehouses and personal clusters, and unlike the 'SQL Editor' it retains previous temporary view definitions in memory so that you can refer to them later on without having to run the entirety of the code.

## Markdown chunks

If you hover between or below a *chunk* two buttons will appear allowing you to add either a markdown block or a code block (`+ Code` or `+ Text`). If you click markdown (`+ Text`) you will get a text box with formatting options.

If for whatever reason you wanted to change a code chunk to a markdown block, you can do this by adding `%md` to the top of the block (for Markdown). Likewise you can change a markdown block to a codeblock of any language using `%r`, `%sql`, `%python` or `%scala`. These are referred to as *magic commands* and they tell Databricks how to interpret the content of each block.


### Headers

Headers in the markdown make the document more readable, but also come with the benefit of making them easier to navigate.

If you click on the top icon, a 'Table of contents' pane opens up and shows you a hierarchical list of headers. You can navigate to any section of the notebook just by clicking on the header.

![Table of contents](../images/notebooks-toc.png)

One thing to bear in mind is that only the first header of a chunk is listed in the table of contents. To make the notebook easiest to navigate each new heading or sub-heading should be created in a new markdown chunk.

This also comes with the benefit of being able to easily drag and drop each section to re-order them if required.

## Code chunks

Although properly formatted documentation is **very cool** (and I'm sure that's exactly what you were all thinking right now), it wouldn't be much use without the code to document.

Like with a markdown chunk you can create a code chunk by clicking the buttons that appear when you hover over the gap between chunks and choosing `+ Code`. You can also convert any chunk to a code block of a given language using the appropriate *magic command* at the top of the chunk.

Once you run a code chunk, any output will be displayed at the bottom of the chunk.


### Default language

As mentioned above, each notebook has a default language, and any chunks that have no language specified by a *magic command* will be assumed to be the default language of the notebook.



### Magic commands

Magic commands allow you to execute code in multiple languages in the same notebook. 

| Magic command | Description |
| --- | --- |
|`%md`| Create a markdown chunk for formatted content|
|`%r`, `%sql`, `%python`, `%scala`| Run code in a cell on a given programming language|
|`%run`| Run script or notebook|

There are some limitations to being able to use several languages in the same notebook though. Each language is isolated from the others and so a variable defined in `python` cannot be accessed in `R` and vice versa.

Data can be shared between the two languages using tables and views by writing out the data in language A and reading it in in language B. Realistically it's probably quite rare that there's a strong use-case for multi-language notebooks, but it doesn't hurt to know how to do it.

It's worth noting that when using a 'SQL warehouse' type of compute you are only able to use SQL so switching to R/python/scala will cause an error.

### Examples

The default language of this notebook is SQL, so I don't need to write any magic commands at the top to execute a SQL query or statement.

`````sql
DECLARE VARIABLE hello_world string;
SET VARIABLE hello_world = 'Hello world!';

SELECT hello_world;
`````

To switch to R we have to use the magic command `%r` at the top.

`````sql
%r
hello_world <- "Hello world!"

print(hello_world)
`````

For python: `%python`

`````sql
%python 
hello_world = "Hello world!"

print(hello_world)
`````


Finally, for scala: `%scala`

`````sql
%scala
def hello_world = "Hello world!"
println(hello_world)
`````

## Widgets

Notebooks can also have [widgets](https://docs.databricks.com/en/notebooks/widgets.html), which are one way of parameterising the code within a notebook. They appear as a text box, or drop down box at the top of a notebook and their contents can be referenced in your code. You can create a widget using any language, and when it is created you must set a default value. 

Widgets are very useful when creating notebooks or processes that may need to be re-run with different values. They can be thought of as similar to arguments in function calls (in R, SQL and python) or stored procedures (SQL).

Below we'll demonstrate the different types of widgets, syntax for SQL and R are provided.

### SQL

We'll first demonstrate using SQL syntax.

#### Text input

The simplest form of widget is simply a free-text field where you can type in what you want it's value to be. Here we'll create a text widget and refer to it in the notebook to pull out it's value.

`````sql
CREATE WIDGET TEXT sql_text DEFAULT 'This is a SQL text widget';
`````

We can now refer to the widget and retrieve it's value.

Run the cell below, then change the contents of the `sql_text` widget and re-run it.

Note that to refer to the widgets value we using the syntax `:widget_name`.

`````sql
SELECT :sql_text;
`````

#### Dropdowns

There may be situations where you want to parameterise something, but only have a limited number of options the user can change it to. For this purpose a dropdown widget is useful.

`````sql
CREATE WIDGET DROPDOWN sql_dropdown DEFAULT 'A' CHOICES SELECT * FROM (VALUES ("A"), ("B"),("C"));
`````

Click on the newly created `sql_dropdown` widget and you'll see the list of options available for a user to select. Choose one of these and then run the chunk below to see your code.

`````sql
SELECT :sql_dropdown;
`````

#### Combobox

For situations where you have many options and want a user to be able to search them using free-text you can use a combobox widget. A combobox will allow a user to search for and select a single value from a list of pre-defined values.



`````sql
CREATE WIDGET COMBOBOX sql_combo DEFAULT "Aadvark" CHOICES SELECT * FROM (VALUES ("Aadvark"), ("Arrow"), ("Apple"),("Banana"));
`````

The chunk above created a drop down widget where you can type and search for one of the options. Delete the word 'Aadvark' from the `sql_combo` widget and you'll see the other available options pop up to be seleted. Choose one and then run the chunk below to see the selected value.

`````sql
SELECT :sql_combo;
`````

#### Multiselect

There may also be situations where you want to allow the user to select several values from a list of predefined values. For this purpose, a multiselect widget can be used.

`````sql
CREATE WIDGET MULTISELECT sql_multi DEFAULT "Aadvark" CHOICES SELECT * FROM (VALUES ("Aadvark"), ("Arrow"), ("Apple"),("Banana"));
`````


In the widget `sql_multi` select a couple of options before running the next cell. 

As with a combobox you can type the value you want to select, and/or select it from the options in the drop down.

`````sql
SELECT :sql_multi;
`````

Here you can see that each selected option is returned, separated by a comma (`,`).

#### Using widgets as catalog / schema / table identifiers

Widgets are quite powerful in that they allow you to parameterise anything in your notebook, including names of catalogs, schemas, and tables or views.

Run the chunk below to create a widget that allows you to select the name of any table in the `steam` schemas of the `catalog_40_copper_longitudinal_ilr` catalog.

`````sql
CREATE WIDGET DROPDOWN sql_table_name DEFAULT 'steam_game_reviews' CHOICES SELECT * FROM (SELECT DISTINCT table_name FROM catalog_40_copper_analyst_training.information_schema.tables WHERE table_schema IN ('steam'));
`````

Select one of the options from the `sql_table_name` widget drop down and then run the cell below.

Below we create a SQL variable called `table_reference` to hold the full path of the table selected in the widget above, then we can use that table reference in a query by wrapping it in the `IDENTIFIER()` function.

After running the cell below choose another option from the drop down and re-run to retrieve the data for a different table.

`````sql
DECLARE OR REPLACE VARIABLE table_reference string;
SET VARIABLE table_reference = CONCAT('catalog_40_copper_analyst_training.steam.',:sql_table_name);

SELECT * FROM IDENTIFIER(table_reference) LIMIT 1000;
`````

### R

Widgets can also be created and referenced in R code, although the syntax is quite different. To create and reference widgets in the R language you need to reference the `dbutils` module which is a library of functions that allows you to interact with Databricks functionality through R/python/scala.  

The `dbutils` module is always available within the Databricks environment, but (like widgets) is not transferable to code outside of Databricks e.g. RStudio.

For more information on the `dbutils` module you can visit the [Databricks Utilities (dbutils) reference](https://docs.databricks.com/en/dev-tools/databricks-utils.html) page in the Databricks documentation.

> Note: The default language for this workbook is `SQL`, so when using `R` commands note the `%r` magic command at the top of each code chunk. This tells Databricks to interpret the code using `R` rather than `SQL`.

#### Textbox

As with the SQL syntax you will need to give the widget a unique name, and a default value when it is created.

`````sql
%r 
dbutils.widgets.text("r_text", "This is an R text widget");
`````

You can store the value of a widget in a variable and use it throughout your code.

`````sql
%r 
r_text <- dbutils.widgets.get("r_text")
r_text
`````

#### Dropdowns

The first two arguments when creating widgets are always the name and the default value. For the choices, these must be supplied as a list of options as a third argument.

`````sql
%r

dbutils.widgets.dropdown("r_dropdown", "A", list("A","B","C"))
`````

`````sql
%r
dbutils.widgets.get("r_dropdown")
`````

#### Combobox

`````sql
%r
dbutils.widgets.combobox("r_combo", "Aadvark", list("Aadvark", "Arrow", "Apple","Banana"))
`````

`````sql
%r

dbutils.widgets.get("r_combo")
`````

#### Multiselect


`````sql
%r
dbutils.widgets.multiselect("r_multi", "Aadvark", list("Aadvark", "Arrow", "Apple","Banana"))
`````

`````sql
%r
dbutils.widgets.get("r_multi")
`````

#### Using widgets as catalog / schema / table identifiers

In R there are a few extra steps to creating a widget that is powered by values in a catalog/database. First we'll have to create a connection using the `sparklyr` library.



`````sql
%r
library(sparklyr)

sc <- sparklyr::spark_connect(method = "databricks")
`````

We can then write a query to get the table names from the `information_schema` of the `catalog_40_copper_analyst_training` catalog, where the schema is `steam`.

> Note: The `sparklyr` package deals with data queries by storing a reference to the data rather than the data itself (similar to a view in SQL). So when you get the results of a query using `sdf_sql()` the dataframe that is returned doesn't actually contain the data until we use the `collect()` function which brings the data into memory.

`````sql
%r
get_schema_table_names <- "SELECT DISTINCT table_name FROM catalog_40_copper_analyst_training.information_schema.tables WHERE table_schema = 'steam';"

table_names <- sdf_sql(sc, get_schema_table_names) %>% collect()
display(table_names)
`````

Now we will need to convert the `table_names` dataframe we fetched above into a list to populate the default and choices options for the widget.

The `as.list()` function turns the `table_names` dataframe column `table_name` into a list which we can pass to the function to create the dropdown widget.

`````sql
%r
table_names_list <- table_names$table_name %>% as.list()
table_names_list
`````

We can now pass the first item of `table_names_list` as the default value for the widget, and the whole list as the options.

`````sql
%r
dbutils.widgets.dropdown("r_table_name", # name of the widget
                        table_names_list[[1]], # default option
                        table_names_list) # choices
`````

We can now retrieve the value of the widget and construct a query to get the tables data before displaying it.

> Note: Here we don't need to use the `IDENTIFIER()` function because we are constructing the query string before we send it to Databricks.

`````sql
%r
table_choice <- dbutils.widgets.get("r_table_name")

sql_query <- paste0("SELECT * FROM catalog_40_copper_analyst_training.steam.",table_choice," LIMIT 1000;")

table_contents <- sdf_sql(sc, sql_query)

display(table_contents)
`````
